# BigQuery Configuration

- Google Cloud Project: careflow-analytics-513310
- Dataset: careflow_dataset
- Table: hospital_event_log
- Total synthetic records: 16
- Unique cases: 5
- Activity types: 5

The BigQuery table contains synthetic hospital activity

## 2. SQL Queries Performed

The following six SQL operations were completed in BigQuery:

0. Create Synthetic Hospital Event Log Table
1. Activity Frequency Analysis
2. Activity Transition Analysis
3. Case-Level Summary Analysis
4. Data Quality Validation Analysis
5. Chronological Event Order Validation

All six queries executed successfully.

```python
results = {
    "Metric": [
        "Total Events",
        "Unique Cases",
        "Unique Activities",
        "Null Case IDs",
        "Null Activity Names",
        "Null Timestamps",
        "Event Order Violations"
    ],
    "Value": [
        16,
        5,
        5,
        0,
        0,
        0,
        0
    ]
}

validation_df = pd.DataFrame(results)

display(validation_df)
```

```python
activity_data = pd.DataFrame({
    "Activity": [
        "Appointment",
        "Prescription",
        "Lab",
        "Imaging",
        "Pharmacy"
    ],
    "Total_Events": [5, 4, 3, 2, 2]
})

plt.figure(figsize=(9, 5))

plt.bar(
    activity_data["Activity"],
    activity_data["Total_Events"]
)

plt.title("Activity Frequency in Synthetic Hospital Event Log")
plt.xlabel("Hospital Activity")
plt.ylabel("Number of Events")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

```python
validation_df.to_csv(
    "careflow_sql_validation_summary.csv",
    index=False
)

activity_data.to_csv(
    "careflow_sql_activity_summary.csv",
    index=False
)

print("Summary CSV files exported successfully!")
```

## 3. Conclusion

The BigQuery SQL analysis was completed successfully
