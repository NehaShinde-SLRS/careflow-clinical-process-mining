```python
import pandas as pd

df = pd.read_csv("HospitalOPD-noname.csv")

df_clean = df.drop(
    columns=["Unnamed: 2"],
    errors="ignore"
).copy()

timestamp_columns = [
    "appointment_start_datetime",
    "prescription_start_datetime",
    "lab_start_datetime",
    "imaging_start_datetime",
    "pharmacy_start_datetime"
]

for col in timestamp_columns:
    df_clean[col] = pd.to_datetime(
        df_clean[col],
        errors="coerce"
    )

print("Dataset loaded and timestamps prepared.")
print("Rows:", len(df_clean))
```

```python
# Compare each service start time with the appointment start time

service_columns = {
    "Prescription": "prescription_start_datetime",
    "Lab": "lab_start_datetime",
    "Imaging": "imaging_start_datetime",
    "Pharmacy": "pharmacy_start_datetime"
}

results = []

for activity, col in service_columns.items():
    valid = df_clean[
        ["appointment_start_datetime", col]
    ].dropna()

    before_count = (
        valid[col] < valid["appointment_start_datetime"]
    ).sum()

    results.append({
        "activity": activity,
        "valid_timestamp_pairs": len(valid),
        "service_before_appointment": int(before_count)
    })

timestamp_summary = pd.DataFrame(results)

display(timestamp_summary)
```

```python
timestamp_summary.to_csv(
    "careflow_timestamp_validation_summary.csv",
    index=False
)

print("Saved careflow_timestamp_validation_summary.csv")
```
